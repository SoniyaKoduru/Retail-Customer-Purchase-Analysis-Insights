```python
# creating a new column of age groups by using pd.qcut function


# assigning the bin values for segmentation
labels = ['young adult','adult','middle-aged','senior']

df['age_group'] = pd.qcut(df['age'],q=4, labels = labels)
df.head()
```

```python
# reviewing the sample of how the age_group is being categorized
df[['age','age_group']].head(10)
```

```python
df['frequency_of_purchases'].unique()
```

```python
# Creating purchase frequency days

freq_days = {'Fortnightly' : 14, 'Weekly' :7, 'Annually' : 365, 'Quarterly' :90, 'Bi-Weekly':14,'Monthly':30, 'Every 3 Months':90 }


df['purchase_frequency_days'] = df['frequency_of_purchases'].map(lambda x : freq_days[x])
```
